```python
def poly_hash(s, base, mod):
    n = len(s)
    h = 0
    for i in range(n):
        h += ord(s[i]) * pow(base, n - 1 - i, mod)
    return h % mod
```

```python
A = input()
B = input()

a = [ord(i) for i in A]
b = [ord(i) for i in B]

if len(b) < len(a):
    print(-1)
else:
    hash_a = poly_hash(A, 32, 256)

    for i in range(0, len(b) - len(a) + 1):
        if poly_hash(B[i:len(a) + i], 32, 256) == hash_a:
            print(i)
    
```
