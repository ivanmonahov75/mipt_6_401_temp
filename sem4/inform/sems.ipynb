```python
def input_sequence(end):
    numbers = []
    
    while True:
        user_input = input()
        
        if user_input == end:
            break
        
        try:
            number = int(user_input)
            numbers.append(number)
        except ValueError:
            print("Ошибка")
    
    return numbers
```

```python
def get_maximums():
    numbers = input_sequence("end")
    for i in range(1, len(numbers) -1 ):
        if numbers[i-1] < numbers[i] and numbers[i] > numbers[i + 1]:
            print(f'{i}th number maximum is {numbers[i]}')
```

```python
get_maximums()
```

```python
def get_sorted_dict():
    numbers = input_sequence("end")
    keys = list(set(numbers))
    vals = []
    for i in keys:
        vals.append(numbers.count(i))
    d = (dict(zip(keys, vals)))
    return {k: v for k, v in sorted(d.items(), key=lambda item: item[1], reverse=True)}
```

```python
get_sorted_dict()
```

```python
def longest_subsequence(increasing=True):
    lst = input_sequence("e")

    max_start = 0
    max_len = 1

    start = 0

    for finish in range(1, len(lst)):
        if (increasing and lst[finish] > lst[finish - 1]) or \
           (not increasing and lst[finish] < lst[finish - 1]):
            continue
        else:
            current_len = finish - start
            if current_len > max_len:
                max_len = current_len
                max_start = start
            start = finish

    # Проверяем последнюю подпоследовательность
    if len(lst) - start > max_len:
        max_len = len(lst) - start
        max_start = start

    return lst[max_start:max_start + max_len], max_len

longest_subsequence()
```

```python
def convert_base(number_str, n, m):
    digits = "0123456789ABCDEF"
    
    if not (2 <= n <= 16 and 2 <= m <= 16):
        return "Ошибка: системы счисления должны быть от 2 до 16."
    
    number_str = number_str.upper()
    
    decimal = 0
    power = 0
    for ch in reversed(number_str):
        decimal += digits.index(ch) * (n ** power)
        power += 1
    
    if decimal == 0:
        return "0"
    
    result = ""
    while decimal > 0:
        remainder = decimal % m
        result = digits[remainder] + result
        decimal //= m
    
    return result

def dec_to_bin(number_str):
    return convert_base(number_str, 10, 2)

def bin_to_dec(number_str):
    return convert_base(number_str, 2, 10)

def dec_to_hex(number_str):
    return convert_base(number_str, 10, 16)

print(dec_to_bin("5"))
```

```python
def is_palindrone(number_str):
    return number_str.upper() == number_str[::-1].upper()

def is_palindromne_bases(number_str):
    for i in range(2, 16 + 1):
        return is_palindrone(convert_base(number_str, 10, 2))

is_palindromne_bases("21")
```

```python
"""
написать функцию, которая возвращает наибольшую из цифр заданного числа в n-ричной системе при его переводе в m-ричную
"""
```

```python
def max_digit_in_base_conversion(number_str, n, m):

    converted = convert_base(number_str, n, m)
    
    
    digits = "0123456789ABCDEF"
    max_digit = '0'
    
    for ch in converted:
        if digits.index(ch) > digits.index(max_digit):
            max_digit = ch
    
    return max_digit
```

```python
"""
написать функцию, которая возвращает вторую по величине цифру заданного числа в n-ричной системе при его переводе в m-ричную
"""
```

```python
def second_max_digit_in_base_conversion(number_str, n, m):
    max_digit = max_digit_in_base_conversion(number_str, n, m)
    converted = convert_base(number_str, n, m)
    converted.replace(max_digit, '')

    digits = "0123456789ABCDEF"
    max_digit = '0'
    
    for ch in converted:
        if digits.index(ch) > digits.index(max_digit):
            max_digit = ch
    
    return max_digit
```

```python
print(second_max_digit_in_base_conversion("123", 10, 16))  # "7B" -> digits: 7,B(11) -> second max: 7 ~ B
```

```python
def get_multipliers(num):
    mult = []
    for i in range(1, num+1):
        if num%i == 0:
            mult.append(i)
    return(mult[::-1])
            
```

```python
get_multipliers(10)
```

```python
def get_num_of_multipliers(num):
    return len(get_multipliers(num))
```

```python
get_num_of_multipliers(10)
```

```python
def is_prime(num):
    return get_num_of_multipliers(num) <= 2
```

```python
is_prime(4)
```

```python
def get_prime_multipliers(num):
    lst = get_multipliers(num)
    for i in lst:
        if not is_prime(i):
            lst.remove(i)
    return lst
```

```python
def is_power_of_prime(num):
    mults = get_multipliers(num)
    if not is_prime(mults[-2]):
        return False
        
    for i in mults[:-2]:
        if not i%mults[-2] == 0:
            return False
    return True
```

```python
is_power_of_prime(5**2)
```

```python
def get_factorization(num):

    fact = [1, ]
    current = num
    continue_factorization = True
    while continue_factorization:
        found_divisor = False
        for i in range(2, current + 1):
            if current % i == 0 and is_prime(i):
                current = current // i  # Use integer division
                fact.append(i)
                found_divisor = True
                break
        
        if not found_divisor or current == 1:
            continue_factorization = False
    
    return fact
```

```python
get_factorization(6)
```

```python
def get_biggests_mult_number(n):
    max_mult = 0
    best_val = None
    for i in range(2, n):
        if max_mult < len(get_multipliers(i)):
            max_mult = len(get_multipliers(i))
            best_val = n
        print(best_val, max_mult, '\n', get_multipliers(i))
    return best_val, max_mult
```
