```python
"""
Создать список чисел меньших 2500, оканчивающихся на 1 и являющихся квадратами целых чисел.
"""
result = []

for n in range(1, int(2500**0.5) + 1):
    square = n * n
    if square < 2500 and square % 10 == 1:
        result.append(square)

print(result)
```

```python
"""
Создать список из 20 случайных чисел от 1 до 100 и вывести:
* все чётные числа
* сумму всех чисел, которые больше 50
"""

my_list = np.random.randint(1, 100, 20)
print("Starting list: ", my_list)
even_numbers = []
for i in my_list:
    if i % 2 == 0:
        even_numbers.append(i)
print("All even numbers: ", np.array(even_numbers), "and there is ", len(even_numbers)," of them")    

max_sum = 0
for i in my_list:
    if i > 50:
        max_sum += i
print("Sum of all numbers grater than 50: ", max_sum)
```

```python
"""
Написать функцию power_list(lst, n), которая получает список чисел и возвращает новый список,
где каждое число возведено в степень n (пример: по запросу power_list([1,2,3], 2) выводится [1, 4, 9])
"""

def power_list(lst, n):
    # return np.power(np.array(lst), n)
    lst_power = []
    for i in lst:
        lst_power.append(i**n)

    return lst_power

print(power_list([1,2,3], 2)) # doesnt work with strings, better to use np array
```

```python
"""
Дан словарь 
"""
grades = {"Alice": 90, "Bob": 75, "Charlie": 82}
"""
* Найти студента с максимальной оценкой
* Посчитать среднюю оценку
"""
max_val = max(grades.values())
for i in grades.keys():
    if grades[i] >= max_val:
        print("Highest score got by: ", i)
        break

print("Average grade: ", round(sum(grades.values())/len(grades), 3))
```

```python
"""
Написать функцию is_leap_year(year), которая проверяет, является ли год високосным.
* Год делится на 4, но не на 100, или делится на 400.
"""
def is_leap_year(year):
    return ((year % 4 == 0) and (year % 100 != 0)) or (year % 400 == 0)

is_leap_year(2026), is_leap_year(2020)
```

```python
"""
Написать программу поиска максимума, минимума, среднего значения и среднеквадратичного отклонения
в последовательности. Числа задаются построчно. Окончание последовательности — слово End.
"""

start = True
lst = []
while(start):
    try:
        lst.append(int(input()))
    except:
        start = False

print("Given list: ", lst)
avg = sum(lst)/len(lst);
print("Max, avg and min value: ", max(lst), " , ", avg, " , ", min(lst))

q_is_sigma = sum((x - avg) ** 2 for x in lst) / (len(lst) - 1)
print("Standart deviation is: ", q_is_sigma)
```

```python
# ввод в листы мы уже преобразовывали, будем делатб сразу для листов 
data = list(map(int, "21 22 23 24 90 25 26 27 120 28".split()))

reg = []
anomaly = []
for val in data:
    
    if len(reg) == 0:
        reg.append(val)
        continue
    dif = val / (sum(reg) / len (reg))
    if dif > 2 or dif < 0.5:
        anomaly.append(dif)
    else:
        reg.append(val)

reg, anomaly
```

```python
sentence = "test, test, testing".split()
num_words = len(sentence)

num_letters = 0
num_signs = 0

max_len = 0
max_word = ""

for word in sentence:
    word_only_letters = ''.join(x for x in word if x.isalpha())
    num_letters += len(word_only_letters)

    num_signs += len(word) - len(word_only_letters)
    
    if len(word_only_letters) > max_len:
        max_word = word_only_letters
        max_len = len(word_only_letters)

avg_len = num_letters / num_words

num_words, num_letters, num_signs, max_word, max_len, avg_len
```
