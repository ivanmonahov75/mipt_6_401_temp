```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
```

```python
with open('data/c11.txt', 'r') as f:
    print(f.read().split('\n'))
```

```python
names = "h20 c1 c2 c3 c4 c5 c6 c7 c8 c9 c10 c11".split()
sigma_data = []
```

```python
for i in names:
    dta = np.loadtxt(f'data/{i}.txt')
    sigma_data.append(np.max(dta) - np.min(dta))
sigma_data.append(sigma_data[-1]*1.02)
k = 73/sigma_data[0]
sigma_data = np.array(sigma_data)*k
```

```python
c_mm = [0.001, 0.018, 0.03, 0.05, 0.08, 0.13, 0.21, 0.34, 0.55, 0.9, 1.125, 1.69, 2.25]
len(c_mm)
```

```python
plt.plot(c_mm, sigma_data, 'o-')
plt.grid()
plt.xlabel(r"$C_{CTAB}$, mM")
plt.ylabel(r"$\sigma$, mN/M")
plt.savefig("img1.png")
```

```python
plt.plot(np.log(c_mm), sigma_data, 'o-')
plt.grid()
plt.xlabel(r"$ln(C)$")
plt.ylabel(r"$\sigma$, mN/M")
k, b = np.polyfit(np.log(c_mm)[-8:-3], sigma_data[-8:-3], 1)
plt.plot(np.log(c_mm)[-8:-2:5], np.log(c_mm)[-8:-2:5]*k + b)
plt.savefig("img2.svg")
k, b
```

```python
def langmuir(c, K):

    return sigma_data[0] - ((5.18*10**-3)*298*8.314)*np.log(1+k*c)
```

```python
g = (5.18*10**-3)*(1-np.exp((-sigma_data[0] + sigma_data)/(8.314*298)))
```

```python
initial_guess = [0.5,]  # Начальное приближение
popt, pcov = curve_fit(langmuir, c_mm, g, p0=initial_guess)
popt
```

```python
plt.plot(c_mm, g)
plt.plot(c_mm, langmuir(np.array(c_mm), popt))
```
