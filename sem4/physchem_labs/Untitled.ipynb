```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
au_data = np.loadtxt("liofobs/IV_COAG.csv", delimiter=',').T
```

```python
import numpy as np
import matplotlib.pyplot as plt

# данные
x = au_data[0]  # длина волны
y = au_data[1]  # оптическая плотность

# поиск максимума
max_idx = np.argmax(y)
x_max = x[max_idx]
y_max = y[max_idx]

plt.figure(figsize=(6,4))

plt.plot(x, y)

# отмечаем максимум
plt.annotate(r'$λ_{max} = ' + f'{x_max:.0f} $нм',
             (x_max, y_max),
             textcoords="offset points",
             xytext=(15,-15))  # было (10,10)

# подписи осей
plt.xlabel('Длина волны, нм')
plt.ylabel('Оптическая плотность, отн. ед.')

plt.grid()
plt.tight_layout()

plt.savefig('au.png')
```

```python

x = np.log10([1, 2, 3])
y = np.log10([1.15e-4, 1.35e-3, 4.68e-2])

plt.figure(figsize=(6,4))
plt.plot(x, y, 'x')
a, b = np.polyfit(x, y, 1)
plt.plot(x[::2], x[::2]*a + b,'-', label=f'y = {round(a, 2)}x + {round(b, 3)}')


plt.xlabel('ln(z)')
plt.ylabel('ln(C)')

plt.grid()
plt.tight_layout()
plt.legend()
plt.savefig('conc.png')
```
