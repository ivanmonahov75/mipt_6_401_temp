```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 1. Данные
# -----------------------------
U = (
    list(range(-100, -1400, -100)) +
    list(range(-1200, 400, 100)) +
    list(range(200, -100, -100))
)

# Данные:
# lower — нижний катет, мм
# upper — верхний катет, мм
# extra:
#   False — обычный угол
#   True  — угол больше 90°, theta = 180 - alpha
data = [
    (39, 46, False),
    (52, 49, False),
    (44, 33, False),
    (39, 19, False),
    (43, 11, False),
    (43, 6, False),
    (38, 11, False),
    (44, 21, False),
    (43, 34, False),
    (32, 44, False),
    (27, 64, False),
    (30, 233, False),
    (15, 113, True),
    (30, 233, False),
    (26, 62, False),
    (59, 81, False),
    (83, 66, False),
    (55, 26, False),
    (66, 19, False),
    (36, 5, False),
    (94, 23, False),
    (51, 24, False),
    (59, 44, False),
    (79, 75, False),
    (63, 74, False),
    (66, 95, False),
    (88, 174, False),
    (87, 283, False),
    (67, 538, False),
    (21, 68, False),
    (31, 61, False),
    (54, 78, False),
]

# -----------------------------
# 2. Константы
# -----------------------------
sigma_Hg_decane = 375      # mN/m
sigma_decane_water = 51    # mN/m

# Погрешность каждого катета, мм
delta_lower = 1
delta_upper = 1

# -----------------------------
# 3. Функция расчета угла и sigma
# -----------------------------
def calc_theta_sigma(lower, upper, extra):
    alpha = np.degrees(np.arctan(upper / lower))

    if extra is True:
        theta = 180 - alpha
    else:
        theta = alpha

    sigma = sigma_Hg_decane + sigma_decane_water * np.cos(np.radians(theta))

    return theta, sigma

# -----------------------------
# 4. Функция расчета погрешности sigma
# -----------------------------
def calc_sigma_error(lower, upper, extra):
    _, sigma_lower_plus = calc_theta_sigma(lower + delta_lower, upper, extra)
    _, sigma_lower_minus = calc_theta_sigma(lower - delta_lower, upper, extra)

    _, sigma_upper_plus = calc_theta_sigma(lower, upper + delta_upper, extra)
    _, sigma_upper_minus = calc_theta_sigma(lower, upper - delta_upper, extra)

    d_sigma_d_lower = (sigma_lower_plus - sigma_lower_minus) / (2 * delta_lower)
    d_sigma_d_upper = (sigma_upper_plus - sigma_upper_minus) / (2 * delta_upper)

    delta_sigma = np.sqrt(
        (d_sigma_d_lower * delta_lower) ** 2 +
        (d_sigma_d_upper * delta_upper) ** 2
    )

    return delta_sigma

# -----------------------------
# 5. Расчет таблицы
# -----------------------------
angles = []
sigmas = []
sigma_errors = []

for lower, upper, extra in data:
    theta, sigma = calc_theta_sigma(lower, upper, extra)
    delta_sigma = calc_sigma_error(lower, upper, extra)

    angles.append(theta)
    sigmas.append(sigma)
    sigma_errors.append(delta_sigma)

df = pd.DataFrame({
    "U, mV": U,
    "lower, mm": [x[0] for x in data],
    "upper, mm": [x[1] for x in data],
    "theta, deg": angles,
    "sigma, mN/m": sigmas,
    "delta_sigma, mN/m": sigma_errors
})

# -----------------------------
# 6. Усреднение по одинаковым U
# -----------------------------
# sigma_mean — среднее значение sigma
# sigma_error — погрешность среднего:
# Δσ_mean = sqrt(sum(Δσ_i^2)) / n

df_mean = (
    df.groupby("U, mV")
      .agg(
          sigma_mean=("sigma, mN/m", "mean"),
          sigma_error=("delta_sigma, mN/m", lambda x: np.sqrt(np.sum(x**2)) / len(x))
      )
      .reset_index()
      .sort_values("U, mV")
)

x = df_mean["U, mV"].values
y = df_mean["sigma_mean"].values
yerr = df_mean["sigma_error"].values

# -----------------------------
# 7. Квадратичная аппроксимация
# y(x) = ax^2 + bx + c
# -----------------------------
coeffs = np.polyfit(x, y, deg=2, w=1 / yerr)
a, b, c = coeffs

# Положение максимума параболы
x_max = -b / (2 * a)
y_max = a * x_max**2 + b * x_max + c

print("Коэффициенты аппроксимации:")
print(f"a = {a:.8e}")
print(f"b = {b:.8e}")
print(f"c = {c:.8f}")
print()
print("Уравнение аппроксимации:")
print(f"y(x) = ({a:.8e})x^2 + ({b:.8e})x + ({c:.8f})")
print()
print("Максимум аппроксимации:")
print(f"x_max = {x_max:.1f} mV")
print(f"y_max = {y_max:.1f} mN/m")

# -----------------------------
# 8. Построение и сохранение графика
# -----------------------------
U_smooth = np.linspace(x.min(), x.max(), 500)
sigma_fit = a * U_smooth**2 + b * U_smooth + c

plt.figure(figsize=(8, 5), dpi=300)

plt.errorbar(
    x,
    y,
    yerr=yerr,
    fmt="o",
    markersize=4,
    capsize=3,
    elinewidth=1,
    linewidth=1,
    label="Experimental data"
)

# Формула аппроксимации для легенды
if b >= 0:
    fit_label = rf"$y(x)={a:.2e}x^2+{b:.2e}x+{c:.2f}$"
else:
    fit_label = rf"$y(x)={a:.2e}x^2-{abs(b):.2e}x+{c:.2f}$"

plt.plot(
    U_smooth,
    sigma_fit,
    linestyle="--",
    linewidth=1.2,
    label=fit_label
)

plt.xlabel(r"$U$, mV", fontsize=12)
plt.ylabel(r"$\sigma$, mN/m", fontsize=12)

plt.grid(True, which="major", linewidth=0.6)
plt.minorticks_on()
plt.grid(True, which="minor", linewidth=0.3, alpha=0.5)

plt.tick_params(
    direction="in",
    which="both",
    top=True,
    right=True
)

plt.legend(frameon=True, fontsize=8)

plt.tight_layout()

# Сохранение графика
plt.savefig("sigma_U_graph.png", dpi=300, bbox_inches="tight")
plt.savefig("sigma_U_graph.pdf", bbox_inches="tight")

plt.show()

# -----------------------------
# 9. Вывод итоговой таблицы
# -----------------------------
df_mean
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 1. Данные: две серии
# -----------------------------

# Серия 1: ход от -100 до -1300 mV
U_1 = list(range(-100, -1400, -100))

data_1 = [
    (39, 46, False),
    (52, 49, False),
    (44, 33, False),
    (39, 19, False),
    (43, 11, False),
    (43, 6, False),
    (38, 11, False),
    (44, 21, False),
    (43, 34, False),
    (32, 44, False),
    (27, 64, False),
    (30, 233, False),
    (15, 113, True),
]

# Серия 2: ход от -1200 до +300 и обратно до 0 mV
U_2 = list(range(-1200, 400, 100)) + list(range(200, -100, -100))

data_2 = [
    (30, 205, False),
    (26, 57, False),
    (59, 74, False),
    (83, 62, False),
    (55, 25, False),
    (66, 18, False),
    (36, 5, False),
    (94, 22, False),
    (51, 23, False),
    (59, 42, False),
    (79, 72, False),
    (63, 70, False),
    (66, 91, False),
    (88, 166, False),
    (87, 270, False),
    (67, 510, False),
    (21, 63, False),
    (31, 57, False),
    (54, 73, False),
]

# -----------------------------
# 2. Константы
# -----------------------------
sigma_Hg_decane = 375      # mN/m
sigma_decane_water = 51    # mN/m

delta_lower = 1            # mm
delta_upper = 1            # mm

# -----------------------------
# 3. Функции расчета
# -----------------------------
def calc_theta_sigma(lower, upper, extra):
    alpha = np.degrees(np.arctan(upper / lower))

    if extra is True:
        theta = 180 - alpha
    else:
        theta = alpha

    sigma = sigma_Hg_decane + sigma_decane_water * np.cos(np.radians(theta))

    return theta, sigma


def calc_sigma_error(lower, upper, extra):
    _, sigma_lower_plus = calc_theta_sigma(lower + delta_lower, upper, extra)
    _, sigma_lower_minus = calc_theta_sigma(lower - delta_lower, upper, extra)

    _, sigma_upper_plus = calc_theta_sigma(lower, upper + delta_upper, extra)
    _, sigma_upper_minus = calc_theta_sigma(lower, upper - delta_upper, extra)

    d_sigma_d_lower = (sigma_lower_plus - sigma_lower_minus) / (2 * delta_lower)
    d_sigma_d_upper = (sigma_upper_plus - sigma_upper_minus) / (2 * delta_upper)

    delta_sigma = np.sqrt(
        (d_sigma_d_lower * delta_lower) ** 2 +
        (d_sigma_d_upper * delta_upper) ** 2
    )

    return delta_sigma


def make_df(U, data, series_name):
    angles = []
    sigmas = []
    sigma_errors = []

    for lower, upper, extra in data:
        theta, sigma = calc_theta_sigma(lower, upper, extra)
        delta_sigma = calc_sigma_error(lower, upper, extra)

        angles.append(theta)
        sigmas.append(sigma)
        sigma_errors.append(delta_sigma)

    df = pd.DataFrame({
        "series": series_name,
        "U, mV": U,
        "lower, mm": [x[0] for x in data],
        "upper, mm": [x[1] for x in data],
        "theta, deg": angles,
        "sigma, mN/m": sigmas,
        "delta_sigma, mN/m": sigma_errors
    })

    return df


def fit_quadratic(df_series):
    x = df_series["U, mV"].values
    y = df_series["sigma, mN/m"].values
    yerr = df_series["delta_sigma, mN/m"].values

    coeffs = np.polyfit(x, y, deg=2, w=1 / yerr)
    a, b, c = coeffs

    x_max = -b / (2 * a)
    y_max = a * x_max**2 + b * x_max + c

    return x, y, yerr, a, b, c, x_max, y_max


def make_fit_label(a, b, c, name):
    if b >= 0:
        return rf"{name}: $y(x)={a:.2e}x^2+{b:.2e}x+{c:.2f}$"
    else:
        return rf"{name}: $y(x)={a:.2e}x^2-{abs(b):.2e}x+{c:.2f}$"


# -----------------------------
# 4. Таблицы
# -----------------------------
df_1 = make_df(U_1, data_1, "Series 1")
df_2 = make_df(U_2, data_2, "Series 2")

df_all = pd.concat([df_1, df_2], ignore_index=True)

# -----------------------------
# 5. Аппроксимации
# -----------------------------
x1, y1, yerr1, a1, b1, c1, x_max1, y_max1 = fit_quadratic(df_1)
x2, y2, yerr2, a2, b2, c2, x_max2, y_max2 = fit_quadratic(df_2)

print("Series 1:")
print(f"y(x) = ({a1:.8e})x^2 + ({b1:.8e})x + ({c1:.8f})")
print(f"x_max = {x_max1:.1f} mV")
print(f"y_max = {y_max1:.1f} mN/m")
print()

print("Series 2:")
print(f"y(x) = ({a2:.8e})x^2 + ({b2:.8e})x + ({c2:.8f})")
print(f"x_max = {x_max2:.1f} mV")
print(f"y_max = {y_max2:.1f} mN/m")

# -----------------------------
# 6. Построение графика
# -----------------------------
U_smooth_1 = np.linspace(min(x1), max(x1), 500)
sigma_fit_1 = a1 * U_smooth_1**2 + b1 * U_smooth_1 + c1

U_smooth_2 = np.linspace(min(x2), max(x2), 500)
sigma_fit_2 = a2 * U_smooth_2**2 + b2 * U_smooth_2 + c2

plt.figure(figsize=(8, 5), dpi=300)

plt.errorbar(
    x1,
    y1,
    yerr=yerr1,
    fmt="o",
    markersize=4,
    capsize=3,
    elinewidth=1,
    linewidth=1,
    label="Series 1"
)

plt.errorbar(
    x2,
    y2,
    yerr=yerr2,
    fmt="s",
    markersize=4,
    capsize=3,
    elinewidth=1,
    linewidth=1,
    label="Series 2"
)

plt.plot(
    U_smooth_1,
    sigma_fit_1,
    linestyle="--",
    linewidth=1.2,
    label=make_fit_label(a1, b1, c1, "Fit 1")
)

plt.plot(
    U_smooth_2,
    sigma_fit_2,
    linestyle=":",
    linewidth=1.5,
    label=make_fit_label(a2, b2, c2, "Fit 2")
)

plt.xlabel(r"$U$, mV", fontsize=12)
plt.ylabel(r"$\sigma$, mN/m", fontsize=12)

plt.grid(True, which="major", linewidth=0.6)
plt.minorticks_on()
plt.grid(True, which="minor", linewidth=0.3, alpha=0.5)

plt.tick_params(
    direction="in",
    which="both",
    top=True,
    right=True
)

plt.legend(frameon=True, fontsize=7)

plt.tight_layout()

plt.savefig("sigma_U_two_series.png", dpi=300, bbox_inches="tight")
plt.savefig("sigma_U_two_series.pdf", bbox_inches="tight")

plt.show()

# -----------------------------
# 7. Итоговые таблицы
# -----------------------------
df_all
```
