```python
import numpy as np
import matplotlib.pyplot as plt
```

```python


# --- Настройки "под ГОСТ" ---
plt.rcParams.update({
    # основной ГОСТ-подобный шрифт
    "font.size": 12,

    "axes.titlesize": 12,
    "axes.labelsize": 12,

    "xtick.labelsize": 10,
    "ytick.labelsize": 10,

    "legend.fontsize": 10,

    "lines.linewidth": 1.5,
    "axes.linewidth": 1.2,

    "grid.linestyle": "--",
    "grid.linewidth": 0.5,
})

# --- Данные (пример) ---
x = np.linspace(0, 10, 100)
y = np.sin(x)

# --- Создание графика ---
fig, ax = plt.subplots(figsize=(6, 4))  # размер в дюймах

ax.plot(x, y, label="sin(x)")

# --- Подписи ---
ax.set_xlabel("x, с")   # обязательно с единицами!
ax.set_ylabel("y, В")
ax.set_title("Зависимость y от x")

# --- Сетка ---
ax.grid(True)

# --- Легенда ---
ax.legend()

# --- Поля (важно для ГОСТ) ---
plt.tight_layout()

# --- Сохранение ---
plt.savefig("figure.png", dpi=300)

plt.show()
```

```python
import os
```

```python
data_files = []
for i in os.listdir():
    if (i[0:5] == "00_00"):
        data_files.append(i)
```

```python
for file in data_files:
    test = (np.loadtxt(file)[::].T)
    plt.plot(test[1], test[2], label=file+f'_{test[2][0]}')
plt.legend()
plt.savefig(f"images/all.png")
plt.show()
```
